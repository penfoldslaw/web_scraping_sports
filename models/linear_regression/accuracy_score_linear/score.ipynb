```python
import pandas as pd
import os

def accuracy_checker(prediction_files, box_score_files, target):

    df = pd.read_csv(prediction_files)
    df[['First', 'Second', 'Third']] = df[target].str.split(' - ', expand=True).astype(int)

    # Check if 'First', 'Second', and 'Third' are all zero, but keep 'First' as is if it is greater than zero
    df['First'] = df.apply(lambda row: 0 if (row['First'] == 0 and row['Second'] == 0 and row['Third'] == 0) else (row['First'] if row['First'] > 0 else 1), axis=1)


    df = df.drop(columns=[f'recentgames_{target}', target])

    # Initialize an empty list to store the dataframes for each box score file
    df_box_list = []

    # Get all files in the box_score_folder_path
    for file_name in os.listdir(box_score_files):
        # Only process files that end with '.csv'
        if file_name.endswith('.csv'):
            file_path = os.path.join(box_score_files, file_name)
            df_box = pd.read_csv(file_path)
            df_box = df_box[['PLAYER', target]]  # Select the necessary columns
            df_box_7 = df_box.head(7)
            next_box = df_box.iloc[15:22]
            df_box = pd.concat([df_box_7, next_box], ignore_index=True)
            # display(df_box)
            df_box_list.append(df_box)  # Append the df_box DataFrame to the list

    # Concatenate all the box score dataframes into one
    df_box_combined = pd.concat(df_box_list, ignore_index=True)

    # Merge the combined box score dataframe with the prediction dataframe
    merged_df = pd.merge(df_box_combined, df, left_on='PLAYER', right_on='Player')
    merged_df = merged_df.dropna()
    merged_df = merged_df[~merged_df['PLAYER'].str.contains('DNP|DND', na=False)]

    # Creates new comparison columns
    # merged_df['first_comparsion'] = (~((merged_df[target] < merged_df['First']) | ((merged_df['First'] == 0) & (merged_df[target] > 0)))).astype(int)
    # merged_df['second_comparsion'] = (~((merged_df[target] < merged_df['Second']) | ((merged_df['Second'] == 0) & (merged_df[target] > 0)))).astype(int)
    # merged_df['third_comparsion'] = (~((merged_df[target] < merged_df['Third']) | ((merged_df['Third'] == 0) & (merged_df[target] > 0)))).astype(int)

    # merged_df['first_comparsion'] = ((merged_df['First'] == 0) & (merged_df[target] >= 0) | (merged_df[target] >= merged_df['First'])).astype(int)
    # merged_df['second_comparsion'] = ((merged_df['Second'] == 0) & (merged_df[target] >= 0) | (merged_df[target] >= merged_df['Second'])).astype(int)
    # merged_df['third_comparsion'] = ((merged_df['Third'] == 0) & (merged_df[target] >= 0) | (merged_df[target] >= merged_df['Third'])).astype(int)


    merged_df['first_comparsion'] = (~((merged_df[target] < merged_df['First']) | ((merged_df['First'] == 0) & (merged_df[target] > 0)))).astype(int)

    merged_df['second_comparsion'] = (~((merged_df[target] < merged_df['Second']) |  ((merged_df['Second'] == 0) & (merged_df[target] > 0)))).astype(int)

    merged_df['third_comparsion'] = (~((merged_df[target] < merged_df['Third']) |   ((merged_df['Third'] == 0) & (merged_df[target] > 0)))).astype(int)

    # Set to 0 when all First, Second, Third are 0 and target > 0
    condition = (merged_df['First'] == 0) & (merged_df['Second'] == 0) & (merged_df['Third'] == 0) & (merged_df[target] > 0)
    merged_df.loc[condition, ['first_comparsion', 'second_comparsion', 'third_comparsion']] = 0








    # merged_df['second_comparsion'] = ((merged_df["Second"] != 0) & (merged_df['Second'] <= merged_df[target])).astype(int)
    # merged_df['third_comparsion'] = ((merged_df["Third"] != 0) & (merged_df['Third'] <= merged_df[target])).astype(int)

    # Calculate the proportion of 1's in each comparison column
    first_comparison_proportion = round(merged_df['first_comparsion'].sum() / len(merged_df) * 100, 2)
    second_comparison_proportion = round(merged_df['second_comparsion'].sum() / len(merged_df) * 100, 2)
    third_comparison_proportion = round(merged_df['third_comparsion'].sum() / len(merged_df) * 100, 2)

    # Print the results
    print(f"Proportion of 1's in first_comparsion: {first_comparison_proportion}")
    print(f"Proportion of 1's in second_comparsion: {second_comparison_proportion}")
    print(f"Proportion of 1's in third_comparsion: {third_comparison_proportion}")

    # Create a dictionary with the calculated proportions
    new_row = {
        'PLAYER': 'Proportions',
        target: 0,
        'Player': '',
        'team': '',
        f'confidence_level_{target}': 0,
        'First': 0,
        'Second': 0,
        'Third': 0,
        'first_comparsion': first_comparison_proportion,
        'second_comparsion': second_comparison_proportion,
        'third_comparsion': third_comparison_proportion
    }

    # Convert the dictionary to a DataFrame with a single row
    new_row_df = pd.DataFrame([new_row])

    # Concatenate the new row to the original DataFrame
    merged_df = pd.concat([merged_df, new_row_df], ignore_index=True)

    pd.set_option('display.max_rows', 1000)  # Maximum number of rows to display
    pd.set_option('display.max_columns', None)  # Show all columns
    pd.set_option('display.width', 1000)  # Adjust column width for better readability


    # Print the updated DataFrame
    return merged_df
```
